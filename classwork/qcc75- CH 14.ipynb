```python
import numpy as np
import matplotlib.pyplot as plt

f = lambda x: np.sin(x)
a, b = 0, 2 * np.pi
N = 50
x = np.linspace(a, b, N)
y = f(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y, 'o-', label='Sampled Points')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Uniform Sampling of f(x) = sin(x)')
plt.grid(True)
plt.legend()
plt.show()

h = (b - a) / (N - 1)
df_dx_central = (f(x + h) - f(x - h)) / (2 * h)
df_dx_forward = (f(x + h) - f(x)) / h
df_dx_backward = (f(x) - f(x - h)) / h
df_dx_true = np.cos(x)
fig, axs = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle("Approximating Derivative of f(x) = sin(x)", fontsize=16)

axs[0, 0].plot(x, df_dx_true, label='Exact Derivative: cos(x)', color='black', linewidth=2)
axs[0, 0].set_title('Exact Derivative')
axs[0, 0].set_xlabel('x')
axs[0, 0].set_ylabel("f'(x)")
axs[0, 0].grid(True)
axs[0, 0].legend()

axs[0, 1].plot(x, df_dx_forward, 'o--', label='Forward Difference', color='red')
axs[0, 1].set_title('Forward Difference')
axs[0, 1].set_xlabel('x')
axs[0, 1].set_ylabel("f'(x)")
axs[0, 1].grid(True)
axs[0, 1].legend()

axs[1, 0].plot(x, df_dx_backward, 'o--', label='Backward Difference', color='blue')
axs[1, 0].set_title('Backward Difference')
axs[1, 0].set_xlabel('x')
axs[1, 0].set_ylabel("f'(x)")
axs[1, 0].grid(True)
axs[1, 0].legend()

axs[1, 1].plot(x, df_dx_central, 'o--', label='Central Difference', color='green')
axs[1, 1].set_title('Central Difference Approximation')
axs[1, 1].set_xlabel('x')
axs[1, 1].set_ylabel("f'(x)")
axs[1, 1].grid(True)
axs[1, 1].legend()

plt.tight_layout()
plt.show()
```
