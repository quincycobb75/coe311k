```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import CubicSpline

x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([0, 1, 0, 1, 0, 1])

# cubic spline
cs = CubicSpline(x, y, bc_type='natural')
x_interp = np.linspace(0, 5, 100)
y_interp = cs(x_interp)

coeffs = np.polyfit(x, y, deg=5)
poly = np.poly1d(coeffs)
y_poly = poly(x_interp)

plt.figure(figsize=(9, 6))
plt.plot(x, y, 'bo', markersize=8, label="Data points", zorder=5)
plt.plot(x_interp, y_interp, 'r-', linewidth=2, label="Cubic Spline (natural)")
plt.plot(x_interp, y_poly, 'g--', linewidth=2, label="Polynomial Fit (deg 5)")
plt.title("Cubic Spline vs Polynomial Fitting", fontsize=14, fontweight='bold')
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('spline_vs_polynomial.png', dpi=150)
plt.show()
```
