```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 2 * np.pi, 50)
```

```python
plt.plot(x, np.sin(x), x, 2*np.sin(x))
plt.show() # Show the graph.
```

```python
plt.plot(x, np.sin(x), "r--", x, 2*np.sin(x))
plt.show() # Show the graph.
```
