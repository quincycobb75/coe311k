```python
import numpy as np
import matplotlib.pyplot as plt

h = 0.0005
t_final = 0.1
t = np.arange(0, t_final, h)
y_euler = np.zeros(len(t))

y_euler[0] = 0

for i in range(0, len(t) - 1):
    dydt = -1000 * y_euler[i] + 3000 - 2000 * np.exp(-t[i])
    y_euler[i+1] = y_euler[i] + h * dydt

y_analytical = 3 - 0.998 * np.exp(-1000 * t) - 2.002 * np.exp(-t)

plt.plot(t, y_euler, 'r--', label="Euler's Forward")
plt.plot(t, y_analytical, 'b', label="Analytical Solution")
plt.legend()
plt.show()
```
