```python
import numpy as np

def is_diagonally_dominant(matrix):
    n = len(matrix)
    
    for i in range(n):
        diagonal_value = abs(matrix[i][i])
        row_sum = 0
        for j in range(n):
            if i != j:
                row_sum += abs(matrix[i][j])
                
        if diagonal_value < row_sum:
            return False
    return True
# example matrix 1
A = np.array([[3,1], 
              [2, 4]])

print(f"Is the matrix diagonally dominant? {is_diagonally_dominant(A)}")
print(A)

# example matrix 2
B = np.array([[5, 6], 
              [7, 8]])

print(f"\nIs the matrix diagonally dominant? {is_diagonally_dominant(B)}")
print(B)
```
