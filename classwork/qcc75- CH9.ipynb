```python
##PART 1 of Challenge
import matplotlib.pyplot as plt
import numpy as np

# Data points
X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T

print ("X: ")
print (A)

print ("Y: ")
print (Y)

print ("A: ")
print (A)

m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}")

plt.scatter(X, Y, label="Data Points")
plt.plot(X, m*X + c, 'r', label="Best Fit")
plt.legend()
plt.show()
```

```python
# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}")
```

# Plot each data point   
# Plot the line y = .95*x + .59

```python
import numpy as np
import matplotlib.pyplot as plt

x_line = np.linspace(min(X), max (X), 100)
y_line = m * x_line

plt.scatter (X, Y, label = "data points")
plt.plot(x_line, y_line, label = "Best fit line: y=mx + b")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Least squares fit")
plt.legend()
plt.grid(True)
plt.show()
```

```python
##PART 2 of Challenge
import matplotlib.pyplot as plt
import numpy as np

X = np.random.rand(10) * 10 
Y = np.random.rand(10) * 10

A = np.vstack([X, np.ones(len(X))]).T
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

print(f"Random Slope: {m:.2f}")
print(f"Random Intercept: {c:.2f}")

plt.figure(figsize=(6, 4))
plt.scatter(X, Y, color='green', label="Random Dots")
plt.plot(X, m*X + c, color='red', label="The Line of Best Fit")
plt.xlabel("x values")
plt.ylabel("v values")
plt.title("Random Vectors")
plt.legend()
plt.grid(True)
plt.show()
```

```python
##PART 3 of Challenge
import matplotlib.pyplot as plt
import numpy as np

# using y=10x+5
X = np.arange(0, 1.01, 0.01)

noise = np.random.normal(0, 0.5, len(X)) 
Y = 10 * X + 5 + noise

A = np.vstack([X, np.ones(len(X))]).T
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

print(f"Your Equation: y = 10x + 5")
print(f"Calculated Fit: y = {m:.2f}x + {c:.2f}")

plt.figure(figsize=(8, 5))
plt.scatter(X, Y, s=15, color='purple', label="Data with Noise")
plt.plot(X, m*X + c, color='black', label=f"Fit Line: y={m:.2f}x+{c:.2f}")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Testing Fit of y = 10x + 5")
plt.legend()
plt.grid(True)
plt.show()
```
