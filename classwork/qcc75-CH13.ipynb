```python
## in class example
import matplotlib.pyplot as plt
import numpy as np

n = np.linspace(1, 100, 100)

plt.plot(n, np.log(n), label='O(log n)')
plt.plot(n, n, label='O(n)')
plt.plot(n, n * np.log(n), label='O(n log n)')
plt.plot(n, n**2, label='O(n^2)')
plt.plot(n, n**3, label='O(n^3)')
plt.plot(n, 2**n, label='O(2^n)')

plt.yscale('log')  # Logarithmic scale for better visualization
plt.title('Big O Complexity Growth Rates')
plt.xlabel('Input Size (n)')
plt.ylabel('Operations')
plt.legend()
plt.show()
```

```python
#CH13
import matplotlib.pyplot as plt
import numpy as np
import time

def naive_multiply(A, B):
    n = len(A)
    result = [[0 for _ in range(n)] for _ in range(n)]
    
    for i in range(n):
        for j in range(n):
            for k in range(n):
                result[i][j] += A[i][k] * B[k][j]
                
    return result

sizes = []
naive_times = []
numpy_times = []

size = 3

for i in range(8):
    sizes.append(size)
    
    A = np.random.rand(size, size)
    B = np.random.rand(size, size)

    A_list = A.tolist()
    B_list = B.tolist()
    
    start = time.time()
    naive_multiply(A_list, B_list)
    end = time.time()
    
    naive_times.append(end - start)
    
    start = time.time()
    np.dot(A, B)
    end = time.time()
    
    numpy_times.append(end - start)
   
    size *= 2

n = range(1, 9)

plt.plot(n, naive_times, label='Naive Multiplication', marker='o')
plt.plot(n, numpy_times, label='NumPy Multiplication', marker='o')

plt.title('Matrix Multiplication Runtime Comparison')
plt.xlabel('Matrix Number (1 to 8)')
plt.ylabel('Runtime (seconds)')
plt.legend()
plt.grid(True)

plt.show()
```
