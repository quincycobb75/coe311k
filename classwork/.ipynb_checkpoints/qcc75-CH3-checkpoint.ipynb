```python
import random

def approximate_pi(num_samples):
    counter = 0
    
    for _ in range(num_samples): # Generate random numbers between -1 and 1
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        
        if x**2 + y**2 < 1:  # Check if the point (x, y) is inside the unit circle
            counter += 1
            
    return 4.0 * counter / num_samples  # Return the ratio of points inside the circle to total points
```

```python
import random
import math

num_runs = 100
num_samples = 1000
true_pi = math.pi

for i in range(1, num_runs + 1):
    counter = 0
    for _ in range(num_samples):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        if (x**2 + y**2) < 1:
            counter += 1
    
    approx_pi = 4.0 * counter / num_samples
    
    abs_error = abs(approx_pi - true_pi) #calculate errors
    rel_error = abs_error / true_pi
  
    print("Run:", i, "| Pi:", approx_pi, "| Error:", abs_error)
```

```python
import random
import numpy

def montecarlo_pi(num_samples):
    number_of_hits = 0
    
    for _ in range(num_samples):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        
        if x * x + y * y < 1:
            number_of_hits += 1
    return 4.0 * number_of_hits / num_samples
```

```python
import math

for i in range (0,100):
        calculated_pi = montecarlo_pi(1000)
        true_pi = numpy.pi
        true_error = abs(true_pi - calculated_pi)
        relative_error = abs(true_pi - calculated_pi)/(true_pi)
        
    print(calculated_pi, true_pi, relative_error)
```
