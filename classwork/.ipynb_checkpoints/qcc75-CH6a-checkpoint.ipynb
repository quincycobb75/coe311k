```python
import numpy as np
import time ##Asked Gemini for help on choosing and inputting time functions from python dictionary

def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    start = time.perf_counter() # Added to track time
    for k in range(max_iterations):
        x_old = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x[j] for j in range(i))
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n))
            x[i] = (b[i] - s1 - s2) / A[i][i] 
        if np.linalg.norm(x_old - x, ord=np.inf) < tol:
            return x, k+1, time.perf_counter() - start
    return x, max_iterations, time.perf_counter() - start

def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    start = time.perf_counter()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
            
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k+1, time.perf_counter() - start
        x = x_new
    return x, max_iterations, time.perf_counter() - start

A = np.array([[10, -1, 2, 0, 0], [-1, 11, -1, 3, 0], [2, -1, 10, -1, 0], [0, 3, -1, 8, -2], [0, 0, 0, -2, 9]])
b = np.array([14, 30, 26, 25, 37])
 
x0 = np.zeros(5)
tol = 1e-6

sol_jac, iter_jac, time_jac = jacobi(A, b, x0, tol, 100)
print(f"Jacobi took {iter_jac} steps and {time_jac:.6f} seconds") ## Runs the Jacobi equation
    
sol_gs, iter_gs, time_gs = gauss_seidel(A, b, x0, tol, 100)
print(f"Gauss-Seidel took {iter_gs} steps and {time_gs:.6f} seconds") ## Runs the Gauss-Seidel equation
```
