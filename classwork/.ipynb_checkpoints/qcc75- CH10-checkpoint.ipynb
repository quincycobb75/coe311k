```python
##Part A
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)
x = np.linspace(-5, 5, 50)
y = 2 * x**2 + 3 * x + np.random.randn(50) * 10

#Linear fit
coef_lin = np.polyfit(x, y, 1)
poly_lin = np.poly1d(coef_lin)
y_pred_lin = poly_lin(x)
mse_lin = np.mean((y - y_pred_lin)**2)
r2_lin = 1 - (np.sum((y - y_pred_lin)**2) / np.sum((y - np.mean(y))**2))

#cubic fit
coef_cub = np.polyfit(x, y, 3)
poly_cub = np.poly1d(coef_cub)
y_pred_cub = poly_cub(x)
mse_cub = np.mean((y - y_pred_cub)**2)
r2_cub = 1 - (np.sum((y - y_pred_cub)**2) / np.sum((y - np.mean(y))**2))

plt.scatter(x, y, color='blue', label='Data points')
plt.plot(x, y_pred_lin, color='green', label='Linear Fit')
plt.plot(x, y_pred_cub, color='red', label='Cubic Fit')
plt.title('Linear vs Cubic')
plt.legend()
plt.show()

print(f"Linear MSE: {mse_lin:.4f}, R-squared: {r2_lin:.4f}")
print(f"Cubic MSE: {mse_cub:.4f}, R-squared: {r2_cub:.4f}")

if mse_cub < mse_lin:
    print("Conclusion: Cubic is the better fit (Lower MSE)")
else:
    print("Conclusion: Linear is the better fit")
```

```python
##Part B
degrees_to_test = [1, 2, 3, 4, 5, 10, 15, 20]

for d in degrees_to_test:
    coeffs = np.polyfit(x, y, d)
    polynomial = np.poly1d(coeffs)
    y_pred = polynomial(x)
    
    mse = np.mean((y - y_pred)**2)
    r_squared = 1 - (np.sum((y - y_pred)**2) / np.sum((y - np.mean(y))**2))
    
    plt.scatter(x, y, color='blue', alpha=0.3)
    plt.plot(x, y_pred, color='red')
    plt.title(f'Degree {d}: MSE={mse:.2f}, R2={r_squared:.4f}')
    plt.show()

    print(f"Degree {d} -> MSE: {mse:.4f}, R2: {r_squared:.4f}")
```

```python
##Part C
degree_v = 3 
V = np.vander(x, N=degree_v + 1, increasing=True)

coeffs_v, residuals, rank, s = np.linalg.lstsq(V, y, rcond=None)
polynomial = np.poly1d(coeffs_v[::-1])
y_pred_v = polynomial(x)

mse_v = np.mean((y - y_pred_v)**2)
ss_tot = np.sum((y - np.mean(y))**2)
ss_res = np.sum((y - y_pred_v)**2)
r2_v = 1 - (ss_res / ss_tot)

plt.scatter(x, y, color='blue', label='Data points')
plt.plot(x, y_pred_v, color='purple', label='Vandermonde Fit')
plt.title('Vandermonde Matrix Fit')
plt.legend()
plt.show()

print("Vandermonde Coefficients:", coeffs_v)
print(f"Vandermonde MSE: {mse_v:.4f}")
print(f"Vandermonde R-squared: {r2_v:.4f}")
```
