```python
import numpy as np
import matplotlib.pyplot as plt

def solve_euler(k, h, x_start=0, x_end=20, y_0=1.0, v_0=0.0):
    """Function to run Euler's method for different parameters."""
    x_values = np.arange(x_start, x_end + h, h)
    y_values = np.zeros(len(x_values))
    v_values = np.zeros(len(x_values))
    
    y_values[0] = y_0
    v_values[0] = v_0 
    
    for i in range(1, len(x_values)):
        x_n = x_values[i - 1]
        y_n = y_values[i - 1]
        v_n = v_values[i - 1]
        
        y_values[i] = y_n + h * v_n
        v_values[i] = v_n + h * (-k * y_n)
        
    return x_values, y_values
    
h_values = [0.01, 0.02, 0.03, 0.05, 0.1, 0.2, 0.3, 0.5]

plt.figure(figsize=(12, 6))
for h in h_values:
    x_val, y_val = solve_euler(k=1.0, h=h, x_end=20)
    plt.plot(x_val, y_val, label=f'h = {h}')

plt.xlabel('x')
plt.ylabel('y(x)')
plt.title("Euler's Method for d²y/dx² = -y with varying step size 'h'")
plt.legend(loc='upper left', bbox_to_anchor=(1, 1))
plt.grid(True)
plt.tight_layout()
plt.savefig('varying_h.png')

k_values = [1.0, 2.0, 5.0, 10.0]

plt.figure(figsize=(12, 6))
for k in k_values:
    x_val, y_val = solve_euler(k=k, h=0.01, x_end=10)
    plt.plot(x_val, y_val, label=f'k = {k}')

plt.xlabel('x')
plt.ylabel('y(x)')
plt.title("Euler's Method with varying 'k' (h = 0.01)")
plt.legend()
plt.grid(True)
plt.savefig('varying_k.png')

test_h_k10 = [0.01, 0.05, 0.1, 0.2]

plt.figure(figsize=(12, 6))
for h in test_h_k10:
    x_val, y_val = solve_euler(k=10.0, h=h, x_end=5)
    plt.plot(x_val, y_val, label=f'h = {h}')

plt.xlabel('x')
plt.ylabel('y(x)')
plt.title("Euler's Method for k = 10 with varying 'h'")
plt.legend()
plt.grid(True)
plt.savefig('k10_varying_h.png')
```

### It is becoming unstable as the x-values increase. It starts to become visually unstable at h=0.01.
### When you start increasing k, the frequency and instability in the graph increase.
### I don't think Euler's Method can solve this with a larger h, as it causes the method to crash and render the graph illegible.
