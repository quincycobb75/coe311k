```python
import numpy as np
import matplotlib.pyplot as plt
import random

def montecarlo_pi(num_samples):
    counter = 0
    
    for i in range(num_samples):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        
    if (x**2 + y**2) <= 1:
        counter += 1   
    return (counter / num_samples) * 4
```

```python
sample_sizes = range(10, 10000, 20)
errors = []
true_pi = np.pi

for num_samples in sample_sizes:
    calc_pi = montecarlo_pi(num_samples)
    errors.append(abs(true_pi - calc_pi))
```

```python
plt.plot(sample_sizes, errors)
plt.xlabel("Number of Samples")
plt.ylabel("Absolute Error")
plt.show() #Asked Gemini AI for help on the plotting which gave me the x and y labeling
```
