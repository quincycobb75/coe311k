```python
import numpy as np
import matplotlib.pyplot as plt
import math

def taylor_sin(x, n_terms):
    approximation = 0
    num = x
    den = 1.0

    for n in range(n_terms):
        sin = (-1)**n
        term = sin * (num / den)
        approximation += term
        
        num = num * (x*x)
        den = den * (2*n + 2) * (2*n + 3)
        
    return approximation

x_values = np.arange(0, 10.1, 0.1)

plt.figure(figsize=(10, 6))

for n in [1, 2, 3, 4]:
    errors = []
    for x in x_values:
        approx = taylor_sin(x, n)
        exact = np.sin(x)
        errors.append(abs(exact - approx)) ##Used Gemini AI to help with this for loop
    
    plt.plot(x_values, errors, label=f'{n} Term(s)')

plt.title('Truncation Error of a sin(x) Taylor Series')
plt.xlabel('x (radians)')
plt.ylabel('Absolute Error')
plt.ylim(0, 10) 
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()  ##The truncation error for x decreases as you move up in terms. The error does not decrease uniformaly as 4 terms is extremely different than the rest. Then the error for x on the small range, 0-4, it is almost virtually non-existant, but as you raise x, it greatly jumps.
```
